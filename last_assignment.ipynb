```python
import requests
from bs4 import BeautifulSoup
import time

#  設定
BASE_URL = "https://www.musashino-u.ac.jp/"
# 処理済みのURLを記録し、同じページを何度も訪れるのを防ぐセット
visited_urls = set()
# サイトマップの結果を格納する辞書 {URL: Title}
sitemap = {}
# 最初の探索対象のURLリスト
urls_to_visit = [BASE_URL]

#  メインのクロール関数
def crawl_website(url_list):
    """
    指定されたURLリストを探索し、同一ドメイン内のリンクを辿る関数
    """
    
    # 探索対象がなくなるまでループ
    while url_list:
        # リストの先頭からURLを取り出す
        current_url = url_list.pop(0)

        # 既に処理済みのURLであればスキップ
        if current_url in visited_urls:
            continue
        
        # 処理済みとして記録
        visited_urls.add(current_url)

        # 負荷軽減のための待機（必須）
        print(f"{current_url}")
        # time.sleep(秒数) でウェブサイトへの負荷を軽減します
        time.sleep(1) 

        try:
            # ウェブサイトにアクセス
            # timeout=10 は、10秒で応答がなければ諦める設定です
            response = requests.get(current_url, timeout=10)
            
            # 成功したステータスコード（200番台）か確認
            if response.status_code != 200:
                print(f"エラー発生: {current_url} (ステータスコード: {response.status_code})")
                continue

            # エンコーディングを正しく設定（文字化けを防ぐ）
            response.encoding = response.apparent_encoding
            
            # BeautifulSoupでHTMLを解析
            soup = BeautifulSoup(response.text, 'html.parser')

            # ページのタイトルを取得
            page_title_tag = soup.find('title')
            
            
            # ページ内の全てのリンクを取得
            # find_all('a', href=True) で、href属性を持つ<a>タグ（リンク）を全て見つける
            for link in soup.find_all('a', href=True):
                href = link['href']
                
                # リンクを絶対URLに変換
                # 例: /about/ を https://.../about/ に変換します
                absolute_url = requests.compat.urljoin(BASE_URL, href)
                
                # 同一ドメイン (BASE_URLから始まる) かつ、まだ処理していないURLのみを対象とする
                if absolute_url.startswith(BASE_URL) and absolute_url not in visited_urls:
                    # 同じURLがリストに重複して入らないようにチェックしてから追加
                    if absolute_url not in url_list:
                        url_list.append(absolute_url)

        except requests.exceptions.RequestException as e:
            # ネットワークエラーなどが発生した場合
            print(f" {current_url} -> {e}")
        except Exception as e:
            # その他の予期せぬエラーが発生した場合
            print(f" {current_url} -> {e}")

#  実行開始
print("--- サイトマップ抽出開始 ---")
crawl_website(urls_to_visit)
print("--- サイトマップ抽出完了 ---")


# 辞書型変数を print() で表示
print(sitemap)
```
